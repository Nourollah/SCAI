```python
import pandas as pd
import plotly.graph_objects as go

# Read the CSV data
data = pd.read_csv('experiment3_mae.csv')

# Ensure data is clean
data = data.dropna()
data['ID'] = data['ID'].astype(float)
data['OOD'] = data['OOD'].astype(float)
models = data['Name'].tolist()

# Collect unique MAE values for y-axis ticks
unique_mae = sorted(set(data['ID'].tolist() + data['OOD'].tolist()))

# Colors
color_id = '#1f77b4'  # Blue for ID
color_ood = '#ff7f0e'  # Orange for OOD
color_line = '#888888'  # Gray for lines and annotations

# Create the dumbbell chart
fig = go.Figure()

# Add lines and points for each model
for i, row in data.iterrows():
    model = row['Name']
    id_mae = row['ID']
    ood_mae = row['OOD']
    # Add connecting line
    fig.add_trace(go.Scatter(
        x=[model, model],
        y=[id_mae, ood_mae],
        mode='lines+markers',
        line=dict(color=color_line, width=2),
        marker=dict(size=10, symbol=['circle', 'square'], color=[color_id, color_ood]),
        name=model,
        showlegend=False
    ))
    # Add difference annotation at midpoint
    diff = ood_mae - id_mae
    mid_y = (id_mae + ood_mae) / 2
    fig.add_annotation(
        x=model,
        y=mid_y,
        text=f'Δ= {diff:.4f}',
        showarrow=False,
        font=dict(size=10, color=color_line),
        yshift=10,
        xshift=15,
        textangle=90
    )

# Add legend traces
fig.add_trace(go.Scatter(
    x=[None], y=[None], mode='markers',
    name='ID MAE', marker=dict(size=10, color=color_id, symbol='circle'),
    legendgroup='ID'
))
fig.add_trace(go.Scatter(
    x=[None], y=[None], mode='markers',
    name='OOD MAE', marker=dict(size=10, color=color_ood, symbol='square'),
    legendgroup='OOD'
))

# Update layout
fig.update_layout(
    title='Vertical Dumbbell Chart: ID vs OOD MAE with Differences',
    xaxis_title='Model',
    yaxis_title='Mean Absolute Error (MAE)',
    yaxis=dict(range=[0, 1], tickformat='.4f'),
    showlegend=True,
    template='plotly_white',
    margin=dict(t=80, b=80),
    xaxis=dict(tickangle=45)
)

# Save to HTML
fig.show()
fig.write_html('dumbbell_with_diff.html')

print("Chart generated and saved as 'dumbbell_with_diff.html'")
```
