```python
data = read('CCCCCCCCCCCCCCC(=O)O_total_xtb.traj', index=':')
```

```python
len(data) / 14
```

```python
# Split the data into 15 equal parts
n = len(data)
split_data = [data[i * n // 14: (i + 1) * n // 14] for i in range(14)]
```

```python
for molecule_set in split_data:
    if molecule_set[0].symbols.formula.format() != molecule_set[-1].symbols.formula.format():
        print(f"Error: {molecule_set[0].symbols.formula.format()} != {molecule_set[-1].symbols.formula.format()}")
```

```python
for molecule_set in split_data:
    write(f'processed/{molecule_set[-1].symbols.formula.format()}.traj', molecule_set)
```
